```python
from datetime import datetime
today = datetime.now().strftime("%Y-%m-%d")
emp_data = {"name":['a','b','c'],"login_time":['9:00','10:00','11:00'], "logout_time":['17:00','18:00','19:00']}
for i in range(len(emp_data['name'])):
    print(f"Employee login details as per Date, login and logout time")
    print(f"Date :{today} - employee_name: {emp_data['name'][i]} - login: {emp_data['login_time'][i]} logout: {emp_data['logout_time'][i]}")
```
