```python
from medical_details import MedicalRequirements, DonateBlood
def get_details():
    patient_name = input("Enter patient name: ")
    age = int(input("Enter patient age: ")) 
    blood_type = input("Enter patient blood type: ")

    medical_req = MedicalRequirements(patient_name, age, blood_type)
    print(f"Patient Name: {medical_req.patient_name}")
    print(f"Age: {medical_req.age}")
    print(f"Blood Type: {medical_req.blood_type}")
    elig_donate_blood(medical_req)
    
    return medical_req

def elig_donate_blood(medical_req):
    if medical_req.age >= 18 and medical_req.age <= 65:
        print(f"{medical_req.patient_name} is eligible to donate blood.")
    else:
        print(f"{medical_req.patient_name} is not eligible to donate blood.")
    

details = get_details()
print(f"Patient Name: {details.patient_name}, Age: {details.age}, Blood Type: {details.blood_type}")

```

```python
from medical_details import MedicalRequirements, DonateBlood
import csv

file_name = input("Enter the CSV file name (with .csv extension): ")
def get_data_from_csv(file_name):
    medical_data = []
    with open(file_name, mode='r') as file:
        csv_reader = csv.DictReader(file)
        for row in csv_reader:
            patient_name = row['Name']
            age = int(row['ExperienceYears'])
            blood_type = 'Unknown'  # Assuming 'blood_type' is a column in your CSV
            medical_req = MedicalRequirements(patient_name, age, blood_type)
            medical_data.append(medical_req)
            print(f"Patient Name: {medical_req.patient_name}, Age: {medical_req.age}, Blood Type: {medical_req.blood_type}")
    return medical_data

Output_data = get_data_from_csv(file_name)
print(f"Total records processed: {len(Output_data)}")

```
