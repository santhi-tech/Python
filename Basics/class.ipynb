```python
class BankAccount:
    def __init__(self, account_number, account_holder, balance=0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}.")
        else:
            print("Deposit amount must be positive.")
            
    def transfer(self, amount, recipient_account):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            recipient_account.balance += amount
            print(f"Transferred ${amount} to {recipient_account.account_holder}. New balance: ${self.balance}.")
        else:
            print("Transfer amount must be positive and less than or equal to the current balance.")


bank_account1 = BankAccount("123456", "Alice", 1000)
bank_account2 = BankAccount("654321", "Bob", 500)

bank_account1.deposit(200)
bank_account1.transfer(300, bank_account2)
balance1 = bank_account1.balance
balance2 = bank_account2.balance
```

```python
class Tutition:
    def __init__(self, name, age, course, fees):
        self.name = name
        self.age = age
        self.course = course
        self.fees = fees
    def student_details(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Course: {self.course}")
        print(f"Fees: ${self.fees}")
    
    def pay_fees(self, amount):
        if amount > 0 and amount <= self.fees:
            self.fees -= amount
            print(f"Paid ${amount}. Remaining fees: ${self.fees}.")
        else:
            print("Payment amount must be positive and less than or equal to the remaining fees.")

student1 = Tutition("John Doe", 20, "Python Programming", 1000)
student1.student_details()
student1.pay_fees(300)
```
